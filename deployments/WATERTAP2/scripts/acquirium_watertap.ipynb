```python
from acquirium import Acquirium

acq = Acquirium(
    server_url="localhost",
    server_port=8000,
    lexicon_path="/Users/umsaka/Documents/DDCPS/acquirium/ontologies/lexicon.json",
)
```

```python
acq.insert_graph("/Users/umsaka/Documents/DDCPS/acquirium/deployments/WATERTAP2/models/test-model.ttl")
```

```python
q = acq.find_entity(_class = "Reverse Osmosis Membrane", alias = "Reverse Osmosis")
q.metadata_head()
```

```python
q2 = acq.find_all_data()
q2.metadata_head()
```

```python
q3 = q2.filter_by_quantity_kind(qk= "Pressure")
q3.metadata_head()
```

```python
q3.add_grafana_panel(
    panel_title="Pressure Data",
    type="TimeSeries"
)
```

```python
q4 = q2.filter_by_substance(substance="Salt-NaCl")
q4.metadata_head()
```
