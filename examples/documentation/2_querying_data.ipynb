# 2 - Querying Data

The code of the tutorial page "2 - Querying Data".
Run `1_inserting_data` first, in the same folder.

```python
import acquirium as aq

acq = aq.init()
```

## Building a query

```python
q = acq.query().measurement()     # every registered stream

q
```

## Reading the data

Wide format, one column per stream:

```python
q.dataframe()
```

Narrow format, one row per reading:

```python
q.dataframe(shape="narrow")
```

## Selecting streams

By label:

```python
acq.query().measurement().where(label="effluent-tds").dataframe()
```

By `source_id` and `stream_name`:

```python
tds = acq.stream("lab-import", "effluent-tds")

acq.query().measurement(stream=tds).dataframe()
```

## Selecting a time range

```python
q.dataframe(
    start="2026-08-08 10:05",
    end="2026-08-08 10:10",
)
```

The latest reading:

```python
q.dataframe(limit=1, order="desc")
```

## Checking what a query selects

```python
q.metadata()
```

## Reading one stream

```python
acq.stream("lab-import", "effluent-tds").dataframe()
```
